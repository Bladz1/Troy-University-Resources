```python
import os
import time
import datetime
import pandas as pd
import numpy as np
import sklearn
import matplotlib.pyplot as plt
import seaborn as sb
from sklearn.linear_model import SGDClassifier
from sklearn.model_selection import train_test_split as ttsplit
from sklearn.model_selection import cross_val_predict as cvpredict
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score
from sklearn.utils import shuffle
```

```python
DataTrain = pd.read_csv("C:\\Users\\Admin\\Documents\\machine learning\\CleanData.csv")
```

```python
DataTrain = pd.get_dummies(DataTrain)
Data = DataTrain.drop(columns='left')
Target = DataTrain['left']
Classifier = SGDClassifier(random_state=30)
```

```python
D_train, D_test, T_train, T_test = ttsplit(Data, Target, test_size=0.15, stratify= Target)
D_train, T_train = shuffle(D_train, T_train)
T_train_pred = cvpredict(Classifier, D_train, T_train, cv=5)
confusion = confusion_matrix(T_train, T_train_pred)
precision = precision_score(T_train, T_train_pred)
recall = recall_score(T_train, T_train_pred)
f1 = f1_score(T_train, T_train_pred)

print("Confusion Matrix: ", confusion)
print("Precision: ", precision)
print("Recall: ", recall)
print("F1 Score: ", f1)
```
