```python
import os
import time
import datetime
import pandas as pd
import numpy as np
import sklearn
import matplotlib.pyplot as plt
import seaborn as sb
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split as ttsplit
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import VotingClassifier
from sklearn.linear_model import Perceptron
```

```python
iris = load_iris()
```

```python
iris
```

```python
DataSet = iris.data[iris.target != 0]
print(DataSet)
DataTarget = iris.target[iris.target != 0]
```

```python
X_train, X_test, y_train, y_test = ttsplit(DataSet, DataTarget, test_size=0.2, random_state=42)
```

```python
svm_clf = SVC(probability=True)
logreg_clf = LogisticRegression()
knn_clf = KNeighborsClassifier()
```

```python
voting_clf = VotingClassifier(
    estimators=[('svm', svm_clf), ('logreg', logreg_clf), ('knn', knn_clf)],
    voting='hard'  # 'hard' for majority voting, 'soft' for weighted voting
)
```

```python
svm_clf.fit(X_train, y_train)
svm_score = svm_clf.score(X_test, y_test)
```

```python
knn_clf.fit(X_train, y_train)
knn_score = knn_clf.score(X_test, y_test)
```

```python
logreg_clf.fit(X_train, y_train)
logreg_score = logreg_clf.score(X_test, y_test)
```

```python
voting_clf.fit(X_train, y_train)
voting_score = voting_clf.score(X_test, y_test)
```

```python
print("SVM Classifier Score:", svm_score)
print("Logistic Regression Classifier Score:", logreg_score)
print("K-Nearest Neighbors Classifier Score:", knn_score)
print("Voting Classifier Score:", voting_score)
```

We can see that Voting score has a higher accuracy than individual accuracy.

```python
perceptron_clf = Perceptron()
perceptron_clf.fit(X_train,y_train)
perceptron_score = perceptron_clf.score(X_test, y_test)
```

```python
print("Perceptron Classifier Score:", perceptron_score)
```

We can see that Perceptron score is not as good as Voting score is.
