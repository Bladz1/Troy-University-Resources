```python
import pandas as pd
from itertools import combinations
```

```python
def load_dataset(num_transactions):
    dataset = []
    for _, row in RawData.head(num_transactions).iterrows():
        transaction = set(row.dropna().values)
        dataset.append(transaction)
    return dataset
```

```python
def get_frequent_itemsets(dataset, support):
    itemsets = {}
    for transaction in dataset:
        for item in transaction:
            itemsets[item] = itemsets.get(item, 0) + 1

    frequent_itemsets = {}
    for item, count in itemsets.items():
        if count >= support:
            frequent_itemsets[frozenset([item])] = count

    return frequent_itemsets
```

```python
def apriori(dataset, support):
    frequent_itemsets = {}
    k = 1

    while True:
        candidates = {}
        for transaction in dataset:
            for itemset in combinations(transaction, k):
                itemset = tuple(sorted(list(itemset)))
                candidates[itemset] = candidates.get(itemset, 0) + 1

        frequent_itemsets_k = {}
        for itemset, count in candidates.items():
            if count >= support:
                frequent_itemsets_k[itemset] = count

        if not frequent_itemsets_k:
            break

        frequent_itemsets.update(frequent_itemsets_k)
        k += 1

    return frequent_itemsets
```

```python
num_transactions = 100
dataset = load_dataset(num_transactions)
dataset
```
